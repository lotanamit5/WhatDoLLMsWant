# Figures for Nir

One place for the figures we show Nir. Each figure has one cell. If a figure is not listed under
**In use**, it is not in this notebook on purpose.

**Two contract paths** (frame = `shopping` everywhere), same end point:
- **Path A** (Graph 1): no contract → `14-inch` → `14-inch + 8GB`
- **Path B** (Graph 1B): no contract → `8GB` → `8GB + 14-inch`
**Rank** = position of the laptop in the model's order, 1 = most preferred. Utilities come from
`FeatureBT` (additive fit: brand + screen + ram). Additive means the brand order is the same inside
every (screen, ram) cell, so brands never cross *within* a spec.

**Style, all figures.** Each requirement has a colour: **14-inch = pink, 8GB = light blue**. A line
segment is coloured by which requirements of the contract it *leads into* the laptop meets:
one → that colour, both → **purple**, none → gray. On path A the first segment (none → `14-inch`) is pink
for 14-inch laptops and gray for the rest, and the second segment (→ `14-inch + 8GB`) is purple /
pink / light blue / gray. On path B the first segment is light blue for 8GB laptops instead. Points use the same rule for the contract they sit on. **Apple laptops are
drawn with a dashed line.** Laptops that meet the full contract have a **bold** label.

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as st
from matplotlib.lines import Line2D
from matplotlib.patches import Patch, Rectangle

from src.bt import FeatureBT

SAVE = True
FIGDIR = "figs/nir"
os.makedirs(FIGDIR, exist_ok=True)

BRANDS = ["ASUS", "Apple", "Dell", "HP", "Lenovo"]
SCREEN, RAM = ["13-inch", "14-inch", "16-inch"], ["4GB", "8GB", "16GB"]
LEVELS = {"brand": BRANDS, "screen": SCREEN, "ram": RAM}
LAPTOPS = [(b, s, r) for b in BRANDS for s in SCREEN for r in RAM]
CONTRACTS = ["none", "screen=14-inch", "ram=8GB", "ram=8GB+screen=14-inch"]   # loaded and fitted
FRAME, EXPECT = "shopping", 9900
MODELS = ([("qwen", s) for s in ["0.5", "7", "32", "72"]]
          + [("gemma", s) for s in ["1", "4", "12", "27"]]
          + [("olmo", s) for s in ["1", "7", "13", "32"]])
HEADLINE = ("qwen", "32")
BASES = ["data/laptops_robustness", "data/laptops_robustness_gemma", "data/laptops_olmo"]
```

```python
def make_path(ids, tag, title):
    """A contract path: ids, each contract as {feature: level}, the order requirements are added,
    and labels written in that order (so path B reads "8GB + 14-inch")."""
    cons = [dict(p.split("=") for p in c.split("+")) if c != "none" else {} for c in ids]
    order = []
    for con in cons:
        order += sorted(set(con) - set(order))
    lab = [" + ".join(con[f] for f in order if f in con) or "no contract" for con in cons]
    short = [l.replace("-inch", "in").replace(" + ", "+").replace("no contract", "none") for l in lab]
    return dict(ids=ids, cons=cons, order=order, lab=lab, short=short, tag=tag, title=title)


PA = make_path(["none", "screen=14-inch", "ram=8GB+screen=14-inch"], "", "path A: 14-inch first")
PB = make_path(["none", "ram=8GB", "ram=8GB+screen=14-inch"], "B", "path B: 8GB first")
FULL = PA["cons"][-1]               # both paths end at the same contract

# One colour per requirement, purple when both are met, gray when none (validated, all pairs).
# Pink / light blue are below 3:1 on the surface, so they always ship with labels or a legend.
PINK, SKY, PURPLE, GRAY = "#e87ba4", "#6da7ec", "#4a3aa7", "#c9c8c4"
REQ_COLOR = {"screen": PINK, "ram": SKY}
MIX = {frozenset(): GRAY, frozenset({"screen"}): PINK, frozenset({"ram"}): SKY,
       frozenset({"screen", "ram"}): PURPLE}
APPLE_LS = (0, (4, 2))  # Apple laptops: dashed line
NEUTRAL = "#8d8b85"     # single colour for Graphs 2 and 3


def met(lvl, con):
    """Which requirements of contract `con` this laptop (or spec) meets."""
    return frozenset(f for f, v in con.items() if lvl.get(f) == v)

SURFACE, INK, INK_MUTED = "#fcfcfb", "#0b0b0b", "#52514e"
plt.rcParams.update({
    "figure.facecolor": SURFACE, "axes.facecolor": SURFACE, "axes.edgecolor": "#d6d5d1",
    "axes.labelcolor": INK, "axes.titlecolor": INK, "text.color": INK,
    "xtick.color": INK_MUTED, "ytick.color": INK_MUTED, "axes.grid": True,
    "grid.color": "#e8e7e3", "axes.axisbelow": True, "axes.spines.top": False,
    "axes.spines.right": False, "figure.dpi": 110, "font.size": 10,
    "savefig.bbox": "tight", "savefig.dpi": 200,
})
```

```python
def load():
    """Complete runs only - config.json without scores.csv is a dead job."""
    R = {}
    for base in BASES:
        for run in sorted(os.listdir(base)):
            cf, sc = (os.path.join(base, run, f) for f in ("config.json", "scores.csv"))
            if not (os.path.isfile(cf) and os.path.isfile(sc)):
                continue
            c = json.load(open(cf))
            key = (c["model_family"], str(c["model_size"]), c["constraints_id"])
            if c["frame"]["name"] != FRAME or key[:2] not in MODELS or key[2] not in CONTRACTS:
                continue
            d = pd.read_csv(sc)
            if len(d) != EXPECT:
                continue
            assert key not in R, f"two complete runs share {key}"
            R[key] = d
    return R


def ranks(d):
    """Rank of each of the 45 laptops, 1 = most preferred."""
    m = FeatureBT(r2_warn=None).fit(d)
    u = pd.Series({l: m.utility(dict(brand=l[0], screen=l[1], ram=l[2])) for l in LAPTOPS})
    return u.rank(ascending=False, method="first")


RUNS = load()
assert len(RUNS) == len(MODELS) * len(CONTRACTS), f"expected {len(MODELS) * len(CONTRACTS)}, got {len(RUNS)}"
RANK = {m: pd.DataFrame({c: ranks(RUNS[(*m, c)]) for c in CONTRACTS}) for m in MODELS}
for R in RANK.values():
    R.index.names = ["brand", "screen", "ram"]
# Mean rank of the 5 laptops that share a spec (screen, ram).
GROUP = {m: R.groupby(level=["screen", "ram"]).mean() for m, R in RANK.items()}


def duel(d):
    """D[x, y] = position-corrected margin of x over y, averaged over templates (> 0: x wins)."""
    t = pd.DataFrame({"A": list(zip(d.a_brand, d.a_screen, d.a_ram)),
                      "B": list(zip(d.b_brand, d.b_screen, d.b_ram)),
                      "m": (d.score_a - d.score_b).values})
    M = t.groupby(["A", "B"])["m"].mean().unstack().reindex(index=LAPTOPS, columns=LAPTOPS)
    return (M - M.T) / 2


def win_ranks(d):
    """Rank by number of head-to-head wins (out of 44); ties broken by the mean margin."""
    D = duel(d)
    key = pd.DataFrame({"wins": (D > 0).sum(axis=1), "margin": D.mean(axis=1)})
    order = key.sort_values(["wins", "margin"], ascending=False).index
    return pd.Series(np.arange(1, len(order) + 1), index=order).reindex(LAPTOPS), key["wins"]


RANK_W, WINS = {}, {}
for m in MODELS:
    out = {c: win_ranks(RUNS[(*m, c)]) for c in CONTRACTS}
    RANK_W[m] = pd.DataFrame({c: out[c][0] for c in CONTRACTS})
    WINS[m] = pd.DataFrame({c: out[c][1] for c in CONTRACTS})
    for X in (RANK_W[m], WINS[m]):
        X.index = pd.MultiIndex.from_tuples(X.index, names=["brand", "screen", "ram"])
print(f"{len(RUNS)} runs, {len(MODELS)} models x {len(CONTRACTS)} contracts")
```

```python
def per_template(d):
    """Fit FeatureBT on each prompt template separately (each one is a full 1980-pair design).

    Templates differ mostly in LOUDNESS (the spread S), not in preference, so each template's
    weights are rescaled to the mean S before comparing. Returns
      W: (feature, level) x template   - rescaled weights
      U: laptop x template             - rescaled utilities (sum of the 3 weights)
      S: raw spread per template
    """
    fits = {t: FeatureBT(r2_warn=None).fit(g) for t, g in d.groupby("template_idx")}
    S = pd.Series({t: f.total_spread() for t, f in fits.items()})
    k = S.mean() / S
    W = pd.DataFrame({t: {(f, l): fits[t].weights_[f][l] * k[t] for f in LEVELS for l in LEVELS[f]}
                      for t in fits})
    W.index.names = ["feature", "level"]
    U = pd.DataFrame({t: {lap: sum(W.loc[(f, v), t] for f, v in zip(["brand", "screen", "ram"], lap))
                          for lap in LAPTOPS} for t in fits})
    U.index.names = ["brand", "screen", "ram"]
    return W, U, S


def item_utility(d):
    """BT with ONE free utility per laptop (45 parameters, no additivity), fitted by OLS on the
    same margins as FeatureBT: m = gamma + u_a - u_b. Centred to mean 0, like the additive one."""
    idx = {l: i for i, l in enumerate(LAPTOPS)}
    a = np.array([idx[t] for t in zip(d.a_brand, d.a_screen, d.a_ram)])
    b = np.array([idx[t] for t in zip(d.b_brand, d.b_screen, d.b_ram)])
    n = len(d)
    X = np.zeros((n, len(LAPTOPS)))
    X[np.arange(n), a] += 1
    X[np.arange(n), b] -= 1
    X = np.column_stack([np.ones(n), X[:, 1:]])          # drop one laptop: it is the reference
    beta = np.linalg.lstsq(X, (d.score_a - d.score_b).values, rcond=None)[0]
    u = np.r_[0.0, beta[1:]]
    return pd.Series(u - u.mean(), index=pd.MultiIndex.from_tuples(LAPTOPS, names=["brand", "screen", "ram"]))


def additive_utility(d):
    f = FeatureBT(r2_warn=None).fit(d)
    return pd.Series([f.utility(dict(brand=l[0], screen=l[1], ram=l[2])) for l in LAPTOPS],
                     index=pd.MultiIndex.from_tuples(LAPTOPS, names=["brand", "screen", "ram"]))


def per_template_item(d):
    """Full parametrization per template: one free utility per laptop (45 params), each template
    rescaled so its spread (best - worst laptop) equals the mean spread - the same scale removal as
    per_template(). Returns laptop x template."""
    I = pd.DataFrame({t: item_utility(g) for t, g in d.groupby("template_idx")})
    rng = I.max() - I.min()
    return I * (rng.mean() / rng)


# Fixed laptop order for the 45-bar grids: by RAM (16 -> 4GB), then screen, then brand.
LAP_ORDER = sorted(LAPTOPS, key=lambda l: (-RAM.index(l[2]), SCREEN.index(l[1]), BRANDS.index(l[0])))


def small_bars(ax, M):
    """Bar = mean over templates, error bar = min-max; no tick labels (used in the grids)."""
    mean = M.mean(axis=1)
    x = np.arange(len(M))
    ax.bar(x, mean, width=0.8, color=NEUTRAL, lw=0, zorder=2)
    ax.errorbar(x, mean, yerr=np.vstack([mean - M.min(axis=1), M.max(axis=1) - mean]), fmt="none",
                ecolor=INK, elinewidth=0.6, capsize=0, zorder=3)
    ax.axhline(0, color=INK_MUTED, lw=0.6, zorder=1)
    ax.grid(axis="x", visible=False)
    ax.set_xlim(-0.8, len(M) - 0.2)
    return x


G3_CONS = ["none", "screen=14-inch", "ram=8GB", "ram=8GB+screen=14-inch"]
G3_LAB = ["no contract", "14-inch", "8GB", "14-inch + 8GB"]


def parse(cid):
    return dict(p.split("=") for p in cid.split("+")) if cid != "none" else {}


def tv_panel(ax, d, cid, n_labels=3, fs=7.5, ms=22):
    """Theoretical (additive) vs empirical (per-laptop) utility for one run. Returns (R2, max gap).

    Points coloured as in Graph 1: by which requirements of THIS run's contract the laptop meets.
    Apple laptops get a thick dark outline."""
    th, em = additive_utility(d), item_utility(d)
    res = em - th
    r2 = 1 - (res ** 2).sum() / ((em - em.mean()) ** 2).sum()
    lim = max(th.abs().max(), em.abs().max()) * 1.1
    ax.plot([-lim, lim], [-lim, lim], color="#d6d5d1", lw=1, zorder=1)
    con = parse(cid)
    lvl = [dict(zip(th.index.names, k)) for k in th.index]
    col = [MIX[met(l, con)] for l in lvl]
    apple = np.array([l["brand"] == "Apple" for l in lvl])
    for sel, ec, lw, z in [(~apple, SURFACE, 0.5, 2), (apple, INK, 1.3, 3)]:
        ax.scatter(th[sel], em[sel], s=ms * (1.25 if z == 3 else 1), c=np.array(col)[sel],
                   edgecolors=ec, linewidths=lw, zorder=z)
    top = res.abs().sort_values(ascending=False).index[:n_labels]
    for i, k in enumerate(sorted(top, key=lambda k: -em[k])):     # stagger: no overlapping labels
        ax.annotate(short(k), (th[k], em[k]), xytext=(18, 10 - 14 * i), textcoords="offset points",
                    fontsize=fs, color=INK, va="center",
                    arrowprops=dict(arrowstyle="-", color=INK_MUTED, lw=0.6))
    ax.set_xlim(-lim, lim); ax.set_ylim(-lim, lim); ax.set_aspect("equal")
    ax.text(0.03, 0.97, f"R$^2$ = {r2:.3f}\nlargest gap = {res.abs().max():.1f}",
            transform=ax.transAxes, va="top", fontsize=fs + 1, color=INK_MUTED)
    return r2, res.abs().max()


def borda_score(d, corrected=True):
    """Raw Borda: number of rows the laptop wins, counted in both slots (no position correction).

    corrected=True: number of the 44 rivals it beats with D > 0, where D = mean of the margins over
    the two slot orders and the 5 templates. The slot bias cancels in D (see Graph 4 notes)."""
    if corrected:
        D = duel(d)
        return (D > 0).sum(axis=1).astype(float).reindex(LAPTOPS)
    win_a = (d.score_a > d.score_b).values
    A = pd.Series(list(zip(d.a_brand, d.a_screen, d.a_ram))); Bs = pd.Series(list(zip(d.b_brand, d.b_screen, d.b_ram)))
    w = pd.Series(np.where(win_a, 1.0, 0.0)).groupby(A.values).sum().add(
        pd.Series(np.where(win_a, 0.0, 1.0)).groupby(Bs.values).sum(), fill_value=0)
    return w.reindex(LAPTOPS)


def rank_desc(x, ties="average"):
    """1 = best. Tied scores share their average rank."""
    return pd.Series(x).rank(ascending=False, method=ties)


BT_TYPES = [("full (one weight per laptop)", item_utility, "#2a78d6", "o"),
            ("additive (feature weights)", additive_utility, "#eb6834", "^")]


def borda_panel(ax, d, ms=18, fs=8, corrected=True):
    """x = Borda rank, y = BT rank, one series per BT type. Returns {type: spearman}."""
    bs = borda_score(d, corrected)
    xr = rank_desc(bs.values)
    out = {}
    ax.plot([0.5, 45.5], [0.5, 45.5], color="#d6d5d1", lw=1, zorder=1)
    for lab, fn, col, mk in BT_TYPES:
        u = fn(d).reindex(LAPTOPS).values
        yr = rank_desc(u, ties="first")
        ax.scatter(xr, yr, s=ms, color=col, marker=mk, edgecolor=SURFACE, lw=0.4, zorder=2, alpha=0.9)
        out[lab] = st.spearmanr(bs.values, u)[0]
    ax.set_xlim(0.5, 45.5); ax.set_ylim(45.5, 0.5); ax.set_aspect("equal")
    ax.set_xticks([1, 15, 30, 45]); ax.set_yticks([1, 15, 30, 45])
    for i, (lab, _, col, _) in enumerate(BT_TYPES):
        ax.text(0.03, 0.03 + 0.07 * (len(BT_TYPES) - 1 - i), f"ρ = {out[lab]:.3f}", transform=ax.transAxes,
                va="bottom", fontsize=fs, color=col, fontweight="bold")
    ax.text(0.97, 0.97, f"{45 - bs.nunique()} tied", transform=ax.transAxes, va="top", ha="right",
            fontsize=fs, color=INK_MUTED)
    return out


def borda_legend(fig, y, fs=8):
    hs = [Line2D([], [], ls="", marker=mk, ms=7, color=col, label=lab) for lab, _, col, mk in BT_TYPES]
    fig.legend(handles=hs, loc="lower left", ncol=2, bbox_to_anchor=(0.01, y), frameon=False, fontsize=fs)


def scatter_legend(fig, y, fs=8):
    hs = [Line2D([], [], ls="", marker="o", ms=7, color=PINK, label="meets 14-inch"),
          Line2D([], [], ls="", marker="o", ms=7, color=SKY, label="meets 8GB"),
          Line2D([], [], ls="", marker="o", ms=7, color=PURPLE, label="meets both"),
          Line2D([], [], ls="", marker="o", ms=7, color=GRAY, label="meets none (or no contract)"),
          Line2D([], [], ls="", marker="o", ms=7, color=GRAY, mec=INK, mew=1.3, label="Apple")]
    fig.legend(handles=hs, loc="lower left", ncol=len(hs), bbox_to_anchor=(0.01, y), frameon=False,
               fontsize=fs, handletextpad=0.3, columnspacing=1.4)


def bars(ax, M, colors, hatches, labels, rot=0):
    """Bar = mean over templates (columns of M), error bar = min-max over templates."""
    mean = M.mean(axis=1)
    err = np.vstack([mean - M.min(axis=1), M.max(axis=1) - mean])
    x = np.arange(len(M))
    ax.bar(x, mean, width=0.78, color=colors, hatch=hatches, edgecolor=SURFACE, lw=0, zorder=2)
    ax.errorbar(x, mean, yerr=err, fmt="none", ecolor=INK, elinewidth=1, capsize=2.5, zorder=3)
    ax.axhline(0, color=INK_MUTED, lw=0.8, zorder=1)
    ax.set_xticks(x, labels, rotation=rot, ha="right" if rot else "center")
    ax.grid(axis="x", visible=False)
    return x


```

```python
def save(fig, name):
    if SAVE:
        fig.savefig(os.path.join(FIGDIR, name + ".png"), facecolor=SURFACE)


def name(m):
    return f"{m[0]}-{m[1]}B"


def short(key):
    *b, s, r = key
    return " ".join(b + [s.replace("-inch", "in"), r])


def bump(ax, R, path, labels=False, fs=7):
    """One line per row of R (columns: contract ids); x = contract on `path`, y = rank (1 on top).

    Segment i -> i+1 is coloured by the requirements of contract i+1 the row meets;
    the point at i by those of contract i. Apple rows are dashed.
    """
    R, CONS = R[path["ids"]], path["cons"]
    x = np.arange(len(CONS))
    ms = 5 if labels else 2.5
    lw = (1.8, 1.0) if labels else (1.2, 0.7)       # (coloured, gray)
    for key, row in R.iterrows():
        lvl = dict(zip(R.index.names, key))
        y = row.values
        apple = lvl.get("brand") == "Apple"
        for i in range(len(CONS) - 1):
            c = MIX[met(lvl, CONS[i + 1])]
            ax.plot(x[i:i + 2], y[i:i + 2], color=c, lw=lw[c == GRAY], zorder=1 if c == GRAY else 2,
                    ls=APPLE_LS if apple else "-", solid_capstyle="round")
        for i in range(len(CONS)):
            c = MIX[met(lvl, CONS[i])]
            ax.plot(x[i], y[i], ls="", marker="o", color=c, ms=ms, mew=0, zorder=3)
        if labels:
            full = met(lvl, CONS[-1]) == frozenset(CONS[-1])
            ink = INK if met(lvl, CONS[-1]) else INK_MUTED
            for xi, ha, off in [(0, "right", -0.06), (x[-1], "left", 0.06)]:
                ax.text(xi + off, row.iloc[xi], short(key), ha=ha, va="center", fontsize=fs,
                        color=ink, fontweight="bold" if full else "normal")
    ax.set_xticks(x, path["lab"])
    pad = 0.62 if labels else 0.08
    ax.set_xlim(-pad, x[-1] + pad)
    ax.set_ylim(len(LAPTOPS) + 0.8, 0.2)
    ax.set_yticks([1, 10, 20, 30, 40, 45])
    ax.grid(axis="x", visible=False)


def contract_legend(fig, path, y=1.0, fs=8, apple=True):
    """Requirement colours (+ the Apple dash) in one row, in the order the path adds them."""
    full = path["cons"][-1]
    hs = [Line2D([], [], color=REQ_COLOR[f], lw=2.2, label=f"meets {full[f]}") for f in path["order"]]
    if len(full) > 1:
        hs += [Line2D([], [], color=PURPLE, lw=2.2, label="meets both")]
    hs += [Line2D([], [], color=GRAY, lw=2.2, label="meets none")]
    if apple:
        hs += [Line2D([], [], color=INK_MUTED, lw=1.8, ls=APPLE_LS, label="Apple")]
    fig.legend(handles=hs, loc="lower left", ncol=len(hs), bbox_to_anchor=(0.02, y),
               frameon=False, fontsize=fs, handlelength=2.2, columnspacing=1.4)


def grid(D, path, title, fname, models=None):
    models = models or MODELS
    nrow = int(np.ceil(len(models) / 4))
    fig, axes = plt.subplots(nrow, 4, figsize=(14, 3.6 * nrow + 0.2), sharey=True, squeeze=False)
    for ax, m in zip(axes.flat, models):
        bump(ax, D[m], path)
        ax.set_title(name(m), loc="left", fontsize=10, fontweight="bold")
        ax.set_xticks(range(len(path["ids"])), path["short"], fontsize=8)
    for ax in axes[:, 0]:
        ax.set_ylabel("rank (1 = most preferred)")
    top = 0.92 if nrow == 3 else 0.89
    fig.tight_layout(rect=(0, 0, 1, top))
    contract_legend(fig, path, y=top + 0.01, fs=9, apple="brand" in D[models[0]].index.names)
    fig.suptitle(f"{title} - {path['title']}", x=0.01, y=1.01, ha="left", fontweight="bold")
    save(fig, fname)
    plt.show()


def single(R, path, title, fname, ylabel="rank (1 = most preferred)", size=(8, 11), fs=7,
           apple=True, top=0.94):
    fig, ax = plt.subplots(figsize=size)
    bump(ax, R, path, labels=True, fs=fs)
    ax.set_ylabel(ylabel)
    fig.tight_layout(rect=(0, 0, 1, top))
    contract_legend(fig, path, y=top + 0.005, apple=apple)
    fig.suptitle(f"{title}, {name(HEADLINE)} - {path['title']}", x=0.02, y=1.0, ha="left",
                 fontweight="bold")
    save(fig, fname)
    plt.show()


# The 9-laptop sample for F1s: one laptop per (screen, RAM) spec, brands spread.
SEED = 0
rng = np.random.default_rng(SEED)
cells = [(s, r) for s in SCREEN for r in RAM]
brands = list(rng.permutation(BRANDS))
SUBSET = [(brands[k % len(brands)], *cells[i]) for k, i in enumerate(rng.permutation(len(cells)))]


def wins_table(path):
    """Wins out of 44 for the 14-inch laptops, and agreement of the two rankings, per contract."""
    from scipy.stats import spearmanr
    rows = []
    for c, l in zip(path["ids"], path["lab"]):
        w = WINS[HEADLINE][c]
        rows.append({"contract": l,
                     "wins 14in 8GB (mean of 5)": w.xs(("14-inch", "8GB"), level=["screen", "ram"]).mean(),
                     "wins 14in 16GB (mean of 5)": w.xs(("14-inch", "16GB"), level=["screen", "ram"]).mean(),
                     "spearman(utility rank, win rank)": spearmanr(RANK[HEADLINE][c], RANK_W[HEADLINE][c])[0]})
    print(pd.DataFrame(rows).round(3).to_string(index=False))


def spec_table(path):
    """Mean rank of the target spec (14-inch, 8GB) and the #1 laptop, per model and contract."""
    rows = []
    for m in MODELS:
        t = GROUP[m].loc[("14-inch", "8GB")]
        top = RANK[m].idxmin()
        rows.append({"model": name(m),
                     **{f"14in+8GB mean rank | {l}": t[c] for c, l in zip(path["ids"], path["lab"])},
                     **{f"#1 | {l}": short(top[c]) for c, l in zip(path["ids"], path["lab"])}})
    pd.set_option("display.width", 240)
    print(pd.DataFrame(rows).round(1).to_string(index=False))
```

# Graph 2 — utility, with the spread across prompt templates

Each of the 5 templates is a full design (all 1980 ordered pairs), so `FeatureBT` is fitted on
each template **separately**. Bar = mean over the 5 fits; error bar = **min–max** over them (5
values are too few for a standard deviation).

**Scale is removed first.** Templates differ mainly in *loudness* (the spread S), not in what the
model prefers, so each template's weights are rescaled to the mean S. Without this the error
bars would mostly show which template is louder. Weights are centred within each feature, so 0 =
the average level, and a laptop's 0 = the average laptop.

All bars are one neutral colour: the contract colours of Graph 1 do not apply here.

```python
G2_MODEL, G2_CON = HEADLINE, "none"
W, U, S_raw = per_template(RUNS[(*G2_MODEL, G2_CON)])

print("raw spread S per template:", S_raw.round(2).to_dict(), f"(max/min = {S_raw.max() / S_raw.min():.2f})")
for nm, M in [("weights (11 levels)", W), ("utilities (45 laptops)", U)]:
    half = (M.max(axis=1) - M.min(axis=1)) / 2
    print(f"{nm}: median half-range {half.median():.2f}, max {half.max():.2f}, "
          f"vs value range {M.mean(axis=1).max() - M.mean(axis=1).min():.2f}")
print("lowest correlation between two templates (utilities):", round(np.corrcoef(U.T.values).min(), 3))
```

**x = position-corrected Borda rank.** For each laptop, count how many of the other 44 it
beats. "Beats" means **D > 0**, where D is its margin against that rival averaged over the two
slot orders and the 5 templates. Every pair is asked in both orders, and the slot bias pushes the
two margins in opposite directions by the same amount, so the average cancels it (T1 shows why
this matters). No model is fitted. Tied laptops share an average rank. This is the win count
behind F1w.

**y = BT rank**, for two BT types: **full** (one weight per laptop) and **additive** (feature
weights, laptop = sum). Both are the OLS-on-margins fits used everywhere in this notebook, and
both remove the slot bias through γ, so x and y compare like with like.

A point on the diagonal = BT ranks that laptop exactly where Borda does. ρ = Spearman between the
Borda score and the BT utility.

Note: on a balanced design like ours, the *classical* BT (logistic on 0/1 wins) ranks identically
to raw Borda up to ties (win totals are its sufficient statistic), so it is left out: it would sit
on the diagonal by construction.

## T1 — why Borda is position-corrected

**Raw Borda** counts every row on its own, so it includes the model's slot habit. In qwen-7B under
`8GB`, all 15 4GB laptops got exactly 70 wins each, because in every 4GB-vs-4GB row the laptop
in **slot B** won: the model did not look at the laptops at all. Raw Borda showed this as one
vertical stack of 15 tied laptops.

The first tables measure the slot habit for every model and contract. For rows where **both
laptops have the same RAM**: the % won by slot B (50 = no bias, 100 = always slot B, 0 = always
slot A), and for 4GB vs 4GB, the % of pairs where **the slot decides** (the winner flips when the
order flips). The last tables compare raw and corrected Borda against BT (numbers only; the raw
figures are no longer drawn).

```python
def slot_bias(d):
    win_b = (d.score_b > d.score_a).values
    out = {}
    for ram in RAM:
        out[f"B wins % | {ram} vs {ram}"] = 100 * win_b[((d.a_ram == ram) & (d.b_ram == ram)).values].mean()
    out["B wins % | different RAM"] = 100 * win_b[(d.a_ram != d.b_ram).values].mean()
    # 4GB vs 4GB: share of unordered pairs (per template) whose winner flips with the order
    t = d[(d.a_ram == "4GB") & (d.b_ram == "4GB")]
    x = t.a_brand + "|" + t.a_screen; y = t.b_brand + "|" + t.b_screen
    key = np.where(x < y, x + "#" + y, y + "#" + x) + "#" + t.template_idx.astype(str).values
    first_wins = np.where(x < y, t.score_a > t.score_b, t.score_b > t.score_a)
    g = pd.Series(first_wins).groupby(key).sum()
    out["4GB vs 4GB: slot decides %"] = 100 * (g == 1).mean()
    return out


T1 = pd.DataFrame([{"model": name(m), "contract": lab, **slot_bias(RUNS[(*m, c)])}
                   for m in MODELS for c, lab in zip(G3_CONS, G3_LAB)])
for col in ["B wins % | 4GB vs 4GB", "4GB vs 4GB: slot decides %"]:
    print(col)
    print(T1.pivot(index="model", columns="contract", values=col).reindex([name(m) for m in MODELS])[G3_LAB].round(1).to_string())
    print()
print("mean over 12 models")
print(T1.groupby("contract", sort=False).mean(numeric_only=True).round(1).to_string())

# raw vs corrected Borda, against both BT types (Spearman), and the tied groups
rows = []
for m in MODELS:
    for c, lab in zip(G3_CONS, G3_LAB):
        d = RUNS[(*m, c)]
        raw, cor = borda_score(d, corrected=False), borda_score(d)
        r = {"model": name(m), "contract": lab}
        for t, fn in [("full", item_utility), ("additive", additive_utility)]:
            u = fn(d).reindex(LAPTOPS).values
            r[f"{t} raw"] = st.spearmanr(raw.values, u)[0]
            r[f"{t} corrected"] = st.spearmanr(cor.values, u)[0]
        r["largest tie raw"] = int(raw.value_counts().max())
        r["largest tie corrected"] = int(cor.value_counts().max())
        rows.append(r)
CMP = pd.DataFrame(rows)
print()
print("raw vs corrected Borda, mean over 12 models")
print(CMP.groupby("contract", sort=False).mean(numeric_only=True).round(3).to_string())
print()
print("largest tied group: raw max", CMP["largest tie raw"].max(), "| corrected max", CMP["largest tie corrected"].max())
```

# Graph 5 — deck-5 figures rebuilt with all 12 models

Three figures from the old deck (`figs4deck5.ipynb`), in their original style, now with the
OLMo models too. The **numbers follow the corrected notebook** (`figs4deck5_fix.ipynb`), not the
original one:

| figure | old source | what is corrected |
|---|---|---|
| F11 adherence | deck 5.5 D | win rates use the position-corrected D (the original counted rows, so the slot decided many of them) |
| F12 feature scales | deck 12.1 | γ = fitted intercept + the PMI constant C (the original plotted γ − C) |
| F13 Apple decay | deck 6.2 | nothing: feature weights are immune to both corrections |

**PMI constant C.** These qwen and gemma runs were collected with `normalize_pmi=True`, so every
stored margin has one constant C subtracted. It is recovered from saturation (largest stored
score per slot), **per run**: C is the same within one collection batch but differs by up to 0.25
between batches. OLMo was collected with PMI off, so C = 0. qwen-0.5B never saturates, so its C,
and its γ, are unknown.

```python
def pmi_offset(runs, tol=0.01, min_sat=0.01):
    """C = logP(L1|base) - logP(L2|base), recovered from saturation (see figs4deck5_fix).
    NaN when the model never saturates."""
    mA = max(d.score_a.max() for d in runs)
    mB = max(d.score_b.max() for d in runs)
    sat = min(max((d.score_a > mA - tol).mean() for d in runs),
              max((d.score_b > mB - tol).mean() for d in runs))
    return (mB - mA if sat >= min_sat else np.nan), sat


# C is estimated PER RUN, not per model: it is identical within one collection batch but
# differs by up to 0.25 between batches (audit 2026-10-07), so pooling a model's runs gave
# some runs the wrong C and moved their gamma by up to 0.25. Weights and D are unaffected.
PMI_C = {}
for m in MODELS:
    for c in G3_CONS:
        PMI_C[(*m, c)] = 0.0 if m[0] == "olmo" else pmi_offset([RUNS[(*m, c)]])[0]   # olmo: PMI off
G5_FITS = {k: FeatureBT(pmi_c=PMI_C[k], r2_warn=None).fit(RUNS[k]) for k in PMI_C}
print(pd.DataFrame({lab: {name(m): PMI_C[(*m, c)] for m in MODELS} for c, lab in zip(G3_CONS, G3_LAB)}).round(3).to_string())


def only_one(d, feat):
    """Rows where a and b differ in `feat` and are tied on every other feature."""
    m = (d[f"a_{feat}"] != d[f"b_{feat}"]).values.copy()
    for o in [f for f in ("brand", "screen", "ram") if f != feat]:
        m &= (d[f"a_{o}"] == d[f"b_{o}"]).values
    return m


def pair_of(d, feat, u, v):
    return (((d[f"a_{feat}"] == u) & (d[f"b_{feat}"] == v))
            | ((d[f"a_{feat}"] == v) & (d[f"b_{feat}"] == u))).values


def paired_D(d, mask, a_is_candidate):
    """One D = u(candidate) - u(other) per unordered pair and template, position-corrected."""
    m = (d.score_a - d.score_b).values
    A = (d.a_brand + "|" + d.a_screen + "|" + d.a_ram).values
    B = (d.b_brand + "|" + d.b_screen + "|" + d.b_ram).values
    first, second = np.where(A < B, A, B), np.where(A < B, B, A)
    key = np.char.add(np.char.add(first.astype(str), "##" + second.astype(str)),
                      "##" + d.template_idx.astype(str).values)
    t = pd.DataFrame({"key": key, "v": np.where(a_is_candidate, m, -m)})[np.asarray(mask)]
    g = t.groupby("key")["v"].agg(["mean", "size"])
    return g[g["size"] == 2]["mean"].values
```

```python
rows = []
for m in MODELS4:
    d1, d0 = RUNS[(*m, "ram=4GB")], RUNS[(*m, "none")]
    w1, w0 = (d1.a_ram == "4GB").values, (d0.a_ram == "4GB").values
    for rival in ("8GB", "16GB"):
        D1 = paired_D(d1, only_one(d1, "ram") & pair_of(d1, "ram", "4GB", rival), w1)
        D0 = paired_D(d0, only_one(d0, "ram") & pair_of(d0, "ram", "4GB", rival), w0)
        rows.append(dict(model=name(m), rival=rival, win_before=100 * (D0 > 0).mean(),
                         win_after=100 * (D1 > 0).mean(), g0=D0.mean(), g1=D1.mean(),
                         kappa=1 - D1.mean() / D0.mean()))
AD4 = pd.DataFrame(rows)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.4), sharey=True)
for ax, rival in zip(axes, ["8GB", "16GB"]):
    t = AD4[AD4.rival == rival].set_index("model").reindex([name(m) for m in MODELS4])
    x = np.arange(len(t))
    ax.bar(x, t.win_after, .55, color=FEATURE_COLOR["ram"], zorder=3)
    for xi, v in zip(x, t.win_after):
        ax.text(xi, v + 1.5, f"{v:.0f}", ha="center", va="bottom", fontsize=8.5, color=INK)
    ax.set_xticks(x, t.index, rotation=30, ha="right")
    ax.set_ylim(0, 112)
    ax.set_title(f"contract asks for 4GB   (4GB vs {rival})", loc="left", fontweight="bold", fontsize=10)
    ax.grid(axis="x", visible=False)
axes[0].set_ylabel("% where 4GB wins")
fig.tight_layout()
save(fig, "F17_adherence_4GB")
plt.show()

print(AD4.set_index(["model", "rival"]).round(2).to_string())
```
